```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

im = Image.open("im1.jpg").convert('RGB')
pixels = np.array(im)

f_transform = np.fft.fft2(pixels, axes=(0, 1))
f_transform_shifted = np.fft.fftshift(f_transform)

magnitude_spectrum = np.log(np.abs(f_transform_shifted) + 1)
magnitude_spectrum = (magnitude_spectrum / np.max(magnitude_spectrum) * 255).astype(np.uint8)
Image.fromarray(magnitude_spectrum).save("out.jpg")

rows, cols, _ = f_transform_shifted.shape
#print(abs(f_transform_shifted)[0])
f_transform_shifted[rows//2:, :, :] = 0
f_transform_shifted[:, cols//2:, :] = 0

f_ishifted = np.fft.ifftshift(f_transform_shifted)
img_back = np.fft.ifft2(f_ishifted, axes=(0, 1))
img_back = np.abs(img_back).astype(np.uint8)

Image.fromarray(img_back).save("out2.jpg")
```

```python
import numpy as np
import scipy.io.wavfile as wav
import matplotlib.pyplot as plt

rate, data = wav.read("in10.wav")
print(rate)

f_transform = np.fft.fft(data, axis=0)
frequencies = np.fft.fftfreq(len(data), 1/rate)

low_freq_noise = 10000 * np.random.normal(0, 0.1, data.shape)
f_transform_with_noise = f_transform + np.fft.fft(low_freq_noise, axis=0)

data_with_noise = np.fft.ifft(f_transform_with_noise, axis=0).real.astype(np.int16)
wav.write("out1.wav", rate, data_with_noise)

cutoff = 1000  # Cutoff frequency in Hz
f_transform_with_noise[np.abs(frequencies) < cutoff, :] = 0

data_without_low_freq = np.fft.ifft(f_transform_with_noise, axis=0).real.astype(np.int16)
wav.write("out2.wav", rate, data_without_low_freq)
```
